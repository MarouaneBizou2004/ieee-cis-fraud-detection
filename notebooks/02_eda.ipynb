# 02 - Exploratory Data Analysis (EDA)

Comprehensive visual exploration of fraud patterns:
1. Transaction amount skewness and fraud distributions
2. Temporal dynamics (hour of day, day of week)
3. Payment card attributes (`card1` to `card6`)
4. Email domain risk profiles
5. Missing value distributions

```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append('..')
from src.data_loader import DataLoader

loader = DataLoader(raw_data_dir='../data/raw')
try:
    df = loader.load_train_data(nrows=50000, reduce_memory=True)
    print('Loaded data shape:', df.shape)
except Exception as e:
    print('Awaiting Kaggle dataset in data/raw.')
```

### 1. Transaction Amount Distribution (Log Scale)
Comparing fraud vs legitimate transactions.

```python
if 'df' in locals():
    plt.figure(figsize=(10, 5))
    sns.kdeplot(data=df[df['isFraud']==0]['TransactionAmt'], label='Legitimate', log_scale=True)
    sns.kdeplot(data=df[df['isFraud']==1]['TransactionAmt'], label='Fraud', log_scale=True, color='red')
    plt.title('Transaction Amount KDE (Log Scale)')
    plt.legend()
    plt.show()
```

### 2. Fraud by Hour of the Day

```python
if 'df' in locals() and 'TransactionDT' in df.columns:
    df['hour'] = (df['TransactionDT'] // 3600) % 24
    hourly_fraud = df.groupby('hour')['isFraud'].mean() * 100
    plt.figure(figsize=(10, 4))
    hourly_fraud.plot(kind='bar', color='#d9534f')
    plt.ylabel('Fraud Rate (%)')
    plt.title('Fraud Rate by Hour of Day (UTC)')
    plt.show()
```

### 3. Fraud by Card Network & Type

```python
if 'df' in locals():
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    df.groupby('card4')['isFraud'].mean().plot(kind='bar', ax=axes[0], color='#2b5c8f')
    axes[0].set_title('Fraud Rate by Card Brand')
    df.groupby('card6')['isFraud'].mean().plot(kind='bar', ax=axes[1], color='#5cb85c')
    axes[1].set_title('Fraud Rate by Card Type')
    plt.show()
```
