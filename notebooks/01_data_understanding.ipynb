This notebook conducts an in-depth data profiling of the Kaggle IEEE-CIS Fraud Detection dataset.
We inspect dataset dimensions, data types, memory footprints, missingness, and the target distribution (`isFraud`).

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

# Add project root to path
sys.path.append('..')
from src.data_loader import DataLoader
from src.utils import reduce_mem_usage, load_config

config = load_config('../config/config.yaml')
loader = DataLoader(raw_data_dir='../data/raw')
status = loader.check_dataset_exists()
print('Raw Dataset Status:', status)
```

### 1. Dataset Dimensions & Inspection
If the full Kaggle dataset is in `data/raw/`, we load the first 50,000 rows for memory-efficient exploration.

```python
try:
    df = loader.load_train_data(nrows=50000, reduce_memory=True)
    summary = loader.get_data_summary(df)
    for k, v in summary.items():
        print(f'{k}: {v}')
except FileNotFoundError as e:
    print('Dataset not found locally. Instructions:')
    print(loader.get_download_instructions())
```

### 2. Target Variable Analysis (`isFraud`)
Analyze class imbalance in fraudulent vs legitimate transactions.

```python
if 'df' in locals() and 'isFraud' in df.columns:
    counts = df['isFraud'].value_counts()
    pcts = df['isFraud'].value_counts(normalize=True) * 100
    print(pd.DataFrame({'Count': counts, 'Percentage (%)': pcts.round(2)}))
```
