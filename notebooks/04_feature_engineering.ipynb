# 04 - Feature Engineering Pipeline

Extracting rich domain-specific signals for fraud detection:
1. **Temporal Features**: `Transaction_hour`, `Transaction_day`
2. **Amount Features**: `TransactionAmt_log`, `TransactionAmt_decimal`
3. **Grouped Card Aggregations**: `TransactionAmt_to_mean_card1`, `TransactionAmt_to_std_card1`
4. **Email Consistency**: Matching purchaser and recipient email domains
5. **Missingness Density**: `null_count` indicator

```python
import sys
import pandas as pd
import numpy as np

sys.path.append('..')
from src.feature_engineering import FeatureEngineer

fe = FeatureEngineer()
sample_tx = pd.DataFrame({
    'TransactionID': [1, 2, 3],
    'TransactionDT': [86400 * 2 + 3600 * 4, 86400 * 5 + 3600 * 14, 86400 * 8 + 3600 * 2],
    'TransactionAmt': [49.99, 1500.00, 25.00],
    'card1': [10020, 10020, 15885],
    'P_emaildomain': ['gmail.com', 'anonymous.com', 'yahoo.com'],
    'R_emaildomain': ['gmail.com', 'protonmail.com', None]
})

fe.fit(sample_tx)
tx_engineered = fe.transform(sample_tx)
print(tx_engineered[['TransactionAmt', 'TransactionAmt_decimal', 'Transaction_hour', 'email_domain_match', 'null_count']])
```
